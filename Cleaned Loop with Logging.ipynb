```python
# Load the data set showing data types, columns , range index and checking the any null value in whole data set


import pandas as pd
import os

# Check files in current directory or find Multi_Touch_Marketingg.csv
files = [f for f in os.listdir('.') if f.endswith('.csv')]
print("CSV files:", files)

# Read the file
df = pd.read_csv('Multi_Touch_Marketingg.csv')
print("\nDataset Info:")
print(df.info())
print("\nFirst 5 rows:")
print(df.head())
```

```python
# Check duplicate records for the same Customer ID, Date, and Touchpoint Order
initial_count = len(df)
duplicate_count = df.duplicated(subset=['Customer_ID', 'Touchpoint_Date', 'Touchpoint_Order']).sum()

print(f"Found {duplicate_count} duplicate records.")

# Drop exact duplicate interactions
df_clean = df.drop_duplicates(subset=['Customer_ID', 'Touchpoint_Date', 'Touchpoint_Order']).copy()

# Sort records chronologically by Customer ID and Touchpoint Order
df_clean = df_clean.sort_values(by=['Customer_ID', 'Touchpoint_Order']).reset_index(drop=True)

print(f"Cleaned dataset shape: {df_clean.shape} (Removed {initial_count - len(df_clean)} rows)")
```

```python
# Convert 'Touchpoint_Date' to standard datetime format (DD-MM-YYYY)
df['Touchpoint_Date'] = pd.to_datetime(df['Touchpoint_Date'], format='%d-%m-%Y', errors='coerce')

# Standardize to UTC timezone
df['Touchpoint_Date_UTC'] = df['Touchpoint_Date'].dt.tz_localize('UTC')

print("Timestamp format standardized to UTC:")
df[['Customer_ID', 'Touchpoint_Date', 'Touchpoint_Date_UTC']].head()
```

```python
import pandas as pd

# Define tracking and UTM parameters to clean
tracking_cols = ['Channel', 'Touchpoint_Type', 'Device']

# 1. Check which columns exist in the DataFrame
existing_cols = [col for col in tracking_cols if col in df.columns]

# 2. Fill missing values across existing tracking columns
df[existing_cols] = df[existing_cols].fillna('Unknown / Direct')

# 3. Verify cleanup and print non-null counts for confirmation
print("=== UTM & Tracking Cleanup Summary ===")
print(df[existing_cols].isnull().sum().to_frame(name="Remaining Nulls"))
print("\nUTM Parameters & Channel fields sanitized successfully.")
```
